## Step 3: Calculate Carbon Footprint and Get Recommendations

**What's happening here?**
When you click the button above, this code will:
1. Connect to real environmental APIs to get current carbon intensity data
2. Calculate your business's total carbon footprint
3. Show you a breakdown by category (electricity, transport, etc.)
4. Generate personalized recommendations based on Islamic values

**Understanding the Results:**
- **Total Emissions**: Your monthly carbon footprint in kg CO2
- **Annual Tonnes**: Your yearly impact (1 tonne = 1,000 kg)
- **Trees Needed**: How many trees it would take to offset your emissions
- **Per Employee**: Impact per person (helps compare with other businesses)

Run this cell, then click the "Calculate" button above!

```python
# Initialize the environmental data processor
processor = EnvironmentalDataProcessor()

# Global variable to store results
footprint_results = None
recommendations = None

def calculate_footprint(button):
    """Calculate carbon footprint when button is clicked."""
    global footprint_results, recommendations
    
    with output_area:
        clear_output()
        print("🔄 Calculating your carbon footprint...")
        print("🌐 Connecting to environmental APIs...")
        
        # Gather business parameters
        business_params = {
            'electricity_kwh_month': electricity_kwh.value,
            'heating_kwh_month': heating_kwh.value,
            'water_m3_month': water_m3.value,
            'waste_kg_month': waste_kg.value,
            'transport_km_month': transport_km.value,
            'transport_mode': transport_mode.value,
            'country': country.value,
            'employees': employees.value
        }
        
        # Calculate footprint
        footprint_results = processor.calculate_business_carbon_footprint(business_params)
        
        # Get recommendations
        recommendations = processor.generate_reduction_recommendations(footprint_results)
        
        # Display results
        print("\n" + "="*70)
        print(f"📊 CARBON FOOTPRINT ANALYSIS FOR: {business_name.value}")
        print("="*70)
        print("")
        
        print(f"🌍 Total Monthly Emissions: {footprint_results['total_emissions_kg_month']:.1f} kg CO2")
        print(f"📅 Annual Emissions: {footprint_results['total_emissions_tonnes_year']:.2f} tonnes CO2")
        print(f"👤 Per Employee: {footprint_results['per_employee_kg_month']:.1f} kg CO2/month")
        print(f"🌳 Trees Needed to Offset: {footprint_results['trees_to_offset_annual']:.0f} trees for 1 year")
        print("")
        
        print("📊 BREAKDOWN BY CATEGORY:")
        print("-" * 70)
        
        breakdown = footprint_results['breakdown']
        percentages = footprint_results['percentages']
        
        categories = [
            ('⚡ Electricity', 'electricity'),
            ('🔥 Heating', 'heating'),
            ('🚚 Transportation', 'transport'),
            ('🗑️ Waste', 'waste'),
            ('💧 Water', 'water')
        ]
        
        for emoji_name, key in categories:
            kg = breakdown[key]
            pct = percentages[key]
            bar = '█' * int(pct / 2)  # Visual bar
            print(f"{emoji_name:20} {kg:8.1f} kg ({pct:5.1f}%) {bar}")
        
        print("")
        print(f"⚡ Current Grid Carbon Intensity: {footprint_results['carbon_intensity']:.0f} g CO2/kWh")
        
        # Create pie chart
        fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
        
        # Pie chart of emissions
        labels = ['Electricity', 'Heating', 'Transport', 'Waste', 'Water']
        sizes = [breakdown['electricity'], breakdown['heating'], breakdown['transport'], 
                breakdown['waste'], breakdown['water']]
        colors = ['#ff9999', '#66b3ff', '#99ff99', '#ffcc99', '#ff99cc']
        explode = (0.1, 0, 0, 0, 0)  # Explode electricity slice
        
        ax1.pie(sizes, explode=explode, labels=labels, colors=colors, autopct='%1.1f%%',
                shadow=True, startangle=90)
        ax1.set_title(f'Carbon Footprint Breakdown\n{business_name.value}', fontsize=14, fontweight='bold')
        
        # Bar chart
        categories_short = ['Electricity', 'Heating', 'Transport', 'Waste', 'Water']
        values = [breakdown['electricity'], breakdown['heating'], breakdown['transport'],
                 breakdown['waste'], breakdown['water']]
        
        bars = ax2.bar(categories_short, values, color=colors, alpha=0.7)
        ax2.set_ylabel('kg CO2 per month', fontsize=12)
        ax2.set_title('Monthly Emissions by Category', fontsize=14, fontweight='bold')
        ax2.tick_params(axis='x', rotation=45)
        
        # Add value labels on bars
        for bar in bars:
            height = bar.get_height()
            ax2.text(bar.get_x() + bar.get_width()/2., height,
                    f'{height:.0f}',
                    ha='center', va='bottom')
        
        plt.tight_layout()
        plt.show()
        
        # Display Islamic-value based recommendations
        print("\n" + "="*70)
        print("🕌 ISLAMIC-VALUE BASED RECOMMENDATIONS")
        print("="*70)
        print("")
        
        for i, rec in enumerate(recommendations, 1):
            priority_emoji = "🔴" if rec['priority'] == 'HIGH' else "🟡" if rec['priority'] == 'MEDIUM' else "🟢"
            print(f"{priority_emoji} {i}. {rec['category']} (Priority: {rec['priority']})")
            print(f"   Impact Potential: {rec['impact_potential']}")
            print(f"   Estimated Reduction: {rec['estimated_reduction']}")
            print("")
            print(f"   📿 Islamic Principle: {rec['islamic_principle']}")
            print("")
            print("   ✅ Recommended Actions:")
            for action in rec['actions']:
                print(f"      • {action}")
            print("")
            print("-" * 70)
            print("")
        
        print("✅ Analysis complete! Scroll down to simulate improvement scenarios.")

# Attach the function to the button
calculate_button.on_click(calculate_footprint)

print("✅ Calculator ready! Fill in your information above and click the button.")
```

## Step 4: Simulate Sustainability Improvements

**What's happening here?**
Now comes the exciting part! We'll use **Monte Carlo simulation** to predict what happens if you make sustainability improvements.

**What is Monte Carlo Simulation?**
Imagine you want to know if reducing electricity by 30% will help. But:
- Some months you might achieve 35% reduction
- Other months only 25% (due to weather, busy periods, etc.)
- Seasonal changes affect energy use
- Unexpected events can happen

Instead of one prediction, we run **10,000 different scenarios** to see:
- Most likely outcome
- Best case scenario
- Worst case scenario
- Average savings over time

**Configure your improvement plan below!**

```python
# Create widgets for intervention scenarios
print("🎯 Configure your sustainability improvement plan:")
print("")

scenario_name = widgets.Text(
    value='Comprehensive Green Plan',
    description='Plan Name:',
    style={'description_width': '200px'},
    layout=widgets.Layout(width='600px')
)

# Reduction targets
electricity_reduction = widgets.FloatSlider(
    value=0.30,
    min=0,
    max=0.80,
    step=0.05,
    description='Electricity Reduction:',
    readout_format='.0%',
    style={'description_width': '200px'},
    layout=widgets.Layout(width='600px')
)

transport_reduction = widgets.FloatSlider(
    value=0.25,
    min=0,
    max=0.70,
    step=0.05,
    description='Transport Reduction:',
    readout_format='.0%',
    style={'description_width': '200px'},
    layout=widgets.Layout(width='600px')
)

waste_reduction = widgets.FloatSlider(
    value=0.40,
    min=0,
    max=0.80,
    step=0.05,
    description='Waste Reduction:',
    readout_format='.0%',
    style={'description_width': '200px'},
    layout=widgets.Layout(width='600px')
)

heating_reduction = widgets.FloatSlider(
    value=0.20,
    min=0,
    max=0.60,
    step=0.05,
    description='Heating Reduction:',
    readout_format='.0%',
    style={'description_width': '200px'},
    layout=widgets.Layout(width='600px')
)

water_reduction = widgets.FloatSlider(
    value=0.30,
    min=0,
    max=0.70,
    step=0.05,
    description='Water Reduction:',
    readout_format='.0%',
    style={'description_width': '200px'},
    layout=widgets.Layout(width='600px')
)

# Implementation timeline
implementation_months = widgets.IntSlider(
    value=6,
    min=1,
    max=24,
    description='Implementation Period (months):',
    style={'description_width': '200px'},
    layout=widgets.Layout(width='600px')
)

# Investment
upfront_cost = widgets.IntText(
    value=5000,
    description='Upfront Investment (£):',
    style={'description_width': '200px'},
    layout=widgets.Layout(width='600px')
)

# Simulation parameters
simulation_months = widgets.IntSlider(
    value=24,
    min=12,
    max=60,
    description='Simulation Duration (months):',
    style={'description_width': '200px'},
    layout=widgets.Layout(width='600px')
)

num_simulations = widgets.IntSlider(
    value=10000,
    min=1000,
    max=50000,
    step=1000,
    description='Monte Carlo Runs:',
    style={'description_width': '200px'},
    layout=widgets.Layout(width='600px')
)

# Simulate button
simulate_button = widgets.Button(
    description='🚀 Run Monte Carlo Simulation',
    button_style='primary',
    layout=widgets.Layout(width='600px', height='50px')
)

simulation_output = widgets.Output()

# Display intervention form
intervention_form = widgets.VBox([
    widgets.HTML("<h3>🎯 Improvement Targets</h3>"),
    scenario_name,
    widgets.HTML("<p><i>Adjust the sliders to set your reduction targets:</i></p>"),
    electricity_reduction,
    transport_reduction,
    waste_reduction,
    heating_reduction,
    water_reduction,
    widgets.HTML("<h3>📅 Implementation Details</h3>"),
    implementation_months,
    upfront_cost,
    widgets.HTML("<h3>⚙️ Simulation Settings</h3>"),
    simulation_months,
    num_simulations,
    widgets.HTML("<br>"),
    simulate_button,
    simulation_output
])

display(intervention_form)
```

## Step 5: Run Simulation and View Results

**What's happening here?**
This is where the magic happens! When you click the button above:

1. **10,000+ scenarios are simulated** with your improvement targets
2. **Realistic variations** are added (seasonal changes, random events, etc.)
3. **Statistical analysis** calculates averages, best/worst cases
4. **Financial metrics** show ROI and payback period

**You'll see:**
- 📈 Emissions over time (with uncertainty bands)
- 💰 Cost savings trajectory
- 🌳 Environmental impact (trees equivalent)
- 💵 Return on investment (ROI)
- ⏱️ Payback period

**Run this cell, then click the Simulate button above!**

```python
def run_simulation(button):
    """Run Monte Carlo simulation when button is clicked."""
    
    with simulation_output:
        clear_output()
        
        # Check if footprint has been calculated
        if footprint_results is None:
            print("❌ Please calculate your carbon footprint first!")
            print("   Go back and click the 'Calculate My Carbon Footprint' button.")
            return
        
        print("🔄 Initializing Monte Carlo simulation...")
        print(f"🎲 Running {num_simulations.value:,} scenarios over {simulation_months.value} months...")
        print("⏳ This may take 10-30 seconds...")
        print("")
        
        # Create simulator
        simulator = SustainabilitySimulator(footprint_results)
        
        # Prepare intervention parameters
        intervention_params = {
            'electricity_reduction': electricity_reduction.value,
            'transport_reduction': transport_reduction.value,
            'waste_reduction': waste_reduction.value,
            'heating_reduction': heating_reduction.value,
            'water_reduction': water_reduction.value,
            'implementation_months': implementation_months.value,
            'upfront_cost': upfront_cost.value
        }
        
        # Run simulation
        results = simulator.simulate_intervention(
            intervention_params,
            months=simulation_months.value,
            num_simulations=num_simulations.value
        )
        
        print("✅ Simulation complete!")
        print("")
        print("="*70)
        print(f"📊 SIMULATION RESULTS: {scenario_name.value}")
        print("="*70)
        print("")
        
        # Environmental Impact
        print("🌍 ENVIRONMENTAL IMPACT:")
        print("-" * 70)
        print(f"📉 Carbon Reduction: {results['percentage_reduction']:.1f}%")
        print(f"🌱 Total CO2 Saved: {results['total_carbon_saved_tonnes']:.2f} tonnes")
        print(f"🌳 Equivalent Trees: {results['trees_equivalent']:.0f} trees for 1 year")
        print(f"📊 Baseline: {results['baseline_monthly']:.1f} kg CO2/month")
        print(f"📊 Final Average: {results['monthly_emissions_mean'][-1]:.1f} kg CO2/month")
        print("")
        
        # Financial Impact
        print("💰 FINANCIAL IMPACT:")
        print("-" * 70)
        print(f"💵 Total Investment: £{upfront_cost.value:,.0f}")
        print(f"💷 Total Savings: £{results['total_cost_savings']:,.2f}")
        print(f"📈 ROI: {results['roi_percent']:.1f}%")
        if results['payback_months']:
            print(f"⏱️ Payback Period: {results['payback_months']} months")
        else:
            print(f"⏱️ Payback Period: Not achieved in simulation period")
        print(f"💰 Net Benefit: £{results['net_benefit'][-1]:,.2f}")
        print("")
        
        # Create visualizations
        months_array = np.arange(1, simulation_months.value + 1)
        
        fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))
        
        # Plot 1: Emissions over time with uncertainty
        ax1.plot(months_array, results['monthly_emissions_mean'], 
                label='Average Emissions', linewidth=2, color='darkblue')
        ax1.fill_between(months_array, 
                        results['monthly_emissions_p10'],
                        results['monthly_emissions_p90'],
                        alpha=0.3, label='10th-90th Percentile', color='lightblue')
        ax1.axhline(y=results['baseline_monthly'], color='red', 
                   linestyle='--', label='Baseline', linewidth=2)
        ax1.set_xlabel('Month', fontsize=12)
        ax1.set_ylabel('CO2 Emissions (kg/month)', fontsize=12)
        ax1.set_title('Carbon Emissions Over Time\n(Monte Carlo Simulation)', 
                     fontsize=14, fontweight='bold')
        ax1.legend()
        ax1.grid(True, alpha=0.3)
        
        # Plot 2: Cumulative savings
        ax2.plot(months_array, results['cumulative_savings'], 
                label='Cumulative Savings', linewidth=2, color='green')
        ax2.plot(months_array, results['cumulative_costs'],
                label='Cumulative Costs', linewidth=2, color='red')
        ax2.fill_between(months_array, 0, results['net_benefit'],
                        where=np.array(results['net_benefit']) > 0,
                        alpha=0.3, color='green', label='Net Profit')
        ax2.fill_between(months_array, 0, results['net_benefit'],
                        where=np.array(results['net_benefit']) <= 0,
                        alpha=0.3, color='red', label='Net Loss')
        ax2.set_xlabel('Month', fontsize=12)
        ax2.set_ylabel('Amount (£)', fontsize=12)
        ax2.set_title('Financial Return Over Time', fontsize=14, fontweight='bold')
        ax2.legend()
        ax2.grid(True, alpha=0.3)
        
        # Plot 3: Emission reduction by category
        categories = ['Electricity', 'Transport', 'Waste', 'Heating', 'Water']
        reductions = [
            electricity_reduction.value * 100,
            transport_reduction.value * 100,
            waste_reduction.value * 100,
            heating_reduction.value * 100,
            water_reduction.value * 100
        ]
        colors_cat = ['#ff9999', '#99ff99', '#ffcc99', '#66b3ff', '#ff99cc']
        
        bars = ax3.barh(categories, reductions, color=colors_cat, alpha=0.7)
        ax3.set_xlabel('Reduction Target (%)', fontsize=12)
        ax3.set_title('Reduction Targets by Category', fontsize=14, fontweight='bold')
        ax3.grid(True, alpha=0.3, axis='x')
        
        # Add value labels
        for i, bar in enumerate(bars):
            width = bar.get_width()
            ax3.text(width, bar.get_y() + bar.get_height()/2,
                    f'{width:.0f}%',
                    ha='left', va='center', fontweight='bold')
        
        # Plot 4: Monthly savings
        ax4.bar(months_array, results['monthly_savings'], 
               color='green', alpha=0.6, label='Monthly Savings')
        ax4.set_xlabel('Month', fontsize=12)
        ax4.set_ylabel('Savings (£)', fontsize=12)
        ax4.set_title('Monthly Cost Savings', fontsize=14, fontweight='bold')
        ax4.legend()
        ax4.grid(True, alpha=0.3, axis='y')
        
        plt.tight_layout()
        plt.show()
        
        # Islamic reflection
        print("="*70)
        print("🕌 ISLAMIC REFLECTION")
        print("="*70)
        print("")
        print("📿 \"And do not commit abuse on the earth, spreading corruption.\"")
        print("   (Quran 2:60)")
        print("")
        print(f"By implementing this plan, you are fulfilling your role as khalifah")
        print(f"(steward) of the Earth. Your {results['percentage_reduction']:.1f}% reduction in emissions")
        print(f"is equivalent to planting {results['trees_equivalent']:.0f} trees.")
        print("")
        print("This is both:")
        print("  ✅ An act of worship (ibadah) - caring for Allah's creation")
        print("  ✅ Financially beneficial - saving £{:,.0f} over {} months".format(
            results['total_cost_savings'], simulation_months.value))
        print("")
        print("May Allah accept your efforts and bless your business. Ameen.")
        print("="*70)

# Attach function to button
simulate_button.on_click(run_simulation)

print("✅ Simulation ready! Configure your plan above and click the button.")
```
